Experiment 2.1:  Write a Python program to perform basic text preprocessing on a given data. The program should:

```python
#Count uppercase letters
uppercase_count = sum(1 for char in text if char.isupper())
print("Number of uppercase letters:", uppercase_count)
```

```python
#1.Convert text to lowercase
text = text.lower()
print(text)
```

```python
#Count different types of punctuation
punctuations = [char for char in text if char in string.punctuation]

different_punctuations = set(punctuations)

print("Different punctuation types:", different_punctuations)
print("Number of different punctuation types:", len(different_punctuations))
```

```python
#2.Remove punctuation
text = text.translate(str.maketrans("", "", string.punctuation))
print(text)
```

```python
#Count Numbers
numbers = re.findall(r'\d+', text)

print("Numbers:", numbers)
print("Number of numbers:", len(numbers))
```

```python
#3.Remove Numbers
text = re.sub(r'\d+', '', text)

print(text)
```

```python
#Count extra white spaces
whitespace = re.findall(r'\s+', text)

extra_whitespace = sum(len(space) - 1 for space in whitespace if len(space) > 1)

print("Amount of extra whitespace:", extra_whitespace)
```

```python
#4.Remove Extra white spaces
text = re.sub(r'\s+', ' ', text).strip()

print(text)
```

```python
#5.Display final cleaned text
print("Cleaned Text:")
print(text)
```

Experiment 2.2: Write a Python program to perform word tokenization and sentence tokenization on a given text. Display the resulting words and sentences separately.

```python
#Read the file
with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()
```

```python
#Without NLTK / spaCy
```

```python
#Read the file
with open("2.2_tokenization_data.txt", "r") as file:
    text = file.read()
```

```python
#Sentence Tokenization
sentences = re.split(r'(?<=[.!?])\s+', text.strip())

print("Sentences:")

for sentence in sentences:
    print(sentence)
```

```python
#Word Tokenization
words = re.findall(r'\b\w+\b', text)

print("Words:")
print(words)
```

```python
print("SENTENCES")

for i, sentence in enumerate(sentences, 1):
    print(i, sentence)

print("WORDS")

for i, word in enumerate(words, 1):
    print(i, word)
```

Experiment 2.3: Write a Python program to tokenize a text and remove commonly occurring stop words such as is, the, a, an, in, of, and, to, etc. Display the original and filtered tokens. (Also, display Stop Words Separately).


```python
#Read the file
with open("2.3_clean_data.txt", "r") as file:
    text = file.read()

print(text)
```

```python
#Tokenize the text
tokens = re.findall(r'\b\w+\b', text.lower())

print("Original Tokens:")
print(tokens)
```

```python
#Define stop words
stop_words = {
    "is", "the", "a", "an", "in", "of", "and", "to",
    "for", "on", "with", "from", "by", "this", "that", "it"
}

print("Stop Words:")
print(stop_words)
```

```python
#Display stop words separately
removed_stop_words = [word for word in tokens if word in stop_words]

print("Stop Words Present in Text:")
print(removed_stop_words)
```

```python
#Remove stop words
filtered_tokens = [word for word in tokens if word not in stop_words]

print("Filtered Tokens:")
print(filtered_tokens)
```

```python
#Display original and filtered token together
print("ORIGINAL TOKENS")
print(tokens)

print("\nSTOP WORDS")
print(removed_stop_words)

print("\nFILTERED TOKENS")
print(filtered_tokens)
```
