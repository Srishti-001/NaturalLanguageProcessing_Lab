**Experiment-4: Term Frequency and NER**

```python
import re
import csv
from collections import Counter
import spacy
```

```python
with open("4.1_4.2_input.txt", "r", encoding="utf-8") as file:
    text = file.read()

print(text)
```

Experimant4.1: To perform Term-Frequency Analysis and Named Entity Recognition on a given text using any toolkit.



●	Calculate the frequency of each word and save the output in CSV format (Term, Frequency)




●	Identify the top 10 most frequent terms and display their corresponding frequencies.

```python
#Preprocess and tokenize
words = re.findall(r'\b\w+\b', text.lower())

print(words[:20])
```

```python
#Calculate Term Frequency
frequency = Counter(words)

print(frequency)
```

```python
#Save frequency in CSV
with open("term_frequency_4_1.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

print("Term frequency saved successfully.")
```

```python
#Display top 10 frequent terms
top_10 = frequency.most_common(10)

print("Top 10 Most Frequent Terms:")
for term, freq in top_10:
    print(term, ":", freq)
```

```python
#Load spaCy model
nlp = spacy.load("en_core_web_sm")

doc = nlp(text)
```

```python
#Perform Named Entity Recognition
print("Named Entities:")

for entity in doc.ents:
    print(entity.text, "->", entity.label_)
```

●	Calculate the frequency of each word and save the output in CSV format (Term, Frequency)

●	Identify the top 10 most frequent terms and display their corresponding frequencies.


```python
#Convert to lowercase
text = text.lower()

print(text[:500])
```

```python
#Remove punctuation
text = re.sub(r'[^\w\s]', '', text)

print(text[:500])
```

```python
#Tokenize
words = text.split()

print(words[:20])
```

```python
#Calculate frequency using dictionary
frequency = {}

for word in words:
    if word in frequency:
        frequency[word] += 1
    else:
        frequency[word] = 1

print(frequency)
```

```python
#Save as CSV
with open("term_frequency_4_2.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)

    writer.writerow(["Term", "Frequency"])

    for term, freq in frequency.items():
        writer.writerow([term, freq])

print("CSV file created successfully.")
```

```python
#Top 10 terms
sorted_frequency = sorted(
    frequency.items(),
    key=lambda x: x[1],
    reverse=True
)

print("Top 10 Most Frequent Terms:")

for term, freq in sorted_frequency[:10]:
    print(term, ":", freq)
```

Experiment4.3: Write a Python program without using NLTK, spaCy, or any other NLP toolkit to:

```python
#1.	create the documents.
documents = [
    "Natural language processing is a field of artificial intelligence.",
    "Natural language processing helps computers understand human language.",
    "Machine learning is an important part of artificial intelligence."
]

print(documents)
```

```python
#Convert to lowercase and remove punctuation
cleaned_documents = []

for document in documents:
    document = document.lower()
    document = re.sub(r'[^\w\s]', '', document)
    cleaned_documents.append(document)

print(cleaned_documents)
```

```python
#Tokenize each document
tokenized_documents = []

for document in cleaned_documents:
    tokens = document.split()
    tokenized_documents.append(tokens)

for i, tokens in enumerate(tokenized_documents):
    print("Document", i + 1, ":", tokens)
```

```python
#Calculate TF
tf = []

for tokens in tokenized_documents:
    word_count = len(tokens)
    frequency = {}

    for word in tokens:
        frequency[word] = frequency.get(word, 0) + 1

    document_tf = {}

    for word, count in frequency.items():
        document_tf[word] = count / word_count

    tf.append(document_tf)

for i, document_tf in enumerate(tf):
    print("\nDocument", i + 1)
    for term, value in document_tf.items():
        print(term, ":", value)
```

```python
#Calculate DF
df = {}

for tokens in tokenized_documents:
    unique_words = set(tokens)

    for word in unique_words:
        df[word] = df.get(word, 0) + 1

print("Document Frequency:")
for term, value in df.items():
    print(term, ":", value)
```

```python
#Calculate IDF
import math

N = len(documents)

idf = {}

for term, document_frequency in df.items():
    idf[term] = math.log(N / document_frequency)

print("Inverse Document Frequency:")
for term, value in idf.items():
    print(term, ":", value)
```

```python
#Calculate TF-IDF
tf_idf = []

for document_tf in tf:
    document_tfidf = {}

    for term, tf_value in document_tf.items():
        document_tfidf[term] = tf_value * idf[term]

    tf_idf.append(document_tfidf)

for i, document_tfidf in enumerate(tf_idf):
    print("\nDocument", i + 1)

    for term, value in document_tfidf.items():
        print(term, ":", value)
```

```python
#Display Term, TF, DF, IDF and TF-IDF
for i, document_tf in enumerate(tf):
    print("\nDocument", i + 1)
    print("-" * 60)
    print("Term\t\tTF\t\tDF\t\tIDF\t\tTF-IDF")

    for term, tf_value in document_tf.items():
        print(
            f"{term:<15}"
            f"{tf_value:.4f}\t\t"
            f"{df[term]}\t\t"
            f"{idf[term]:.4f}\t\t"
            f"{tf_idf[i][term]:.4f}"
        )
```

```python
#Top 10 TF-IDF terms for each document
for i, document_tfidf in enumerate(tf_idf):

    top_terms = sorted(
        document_tfidf.items(),
        key=lambda x: x[1],
        reverse=True
    )[:10]

    print("\nTop 10 TF-IDF Terms - Document", i + 1)

    for term, score in top_terms:
        print(term, ":", round(score, 4))
```
