**Experiment 1 : Employee Data Analysis**

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("employee_information_100.csv")
```

```python
df.head()
```

Experiment 1.1: Read the employee dataset and display the average salary of each department using a bar chart.

```python
# Find average salary department-wise
avg_salary = df.groupby("Department")["Salary"].mean()

print(avg_salary)

# Draw bar chart
plt.figure(figsize=(8,5))
avg_salary.plot(kind="bar", color="skyblue")

plt.title("Average Salary of Each Department")
plt.xlabel("Department")
plt.ylabel("Average Salary")
plt.grid(axis="y")
plt.show()
```

Experiment 1.2: Display the number of employees working in each department.

```python
# Count employees in every department
dept_count = df["Department"].value_counts()

print(dept_count)

# Draw bar chart
plt.figure(figsize=(8,5))
dept_count.plot(kind="bar", color="orange")

plt.title("Number of Employees in Each Department")
plt.xlabel("Department")
plt.ylabel("Employee Count")
plt.grid(axis="y")
plt.show()
```

Experiment 1.3: Display the percentage of male and female employees using a pie chart.

```python
# Count male and female employees
gender_count = df["Gender"].value_counts()

print(gender_count)

# Draw pie chart
plt.figure(figsize=(6,6))
plt.pie(gender_count,
        labels=gender_count.index,
        autopct="%1.1f%%",
        startangle=90)

plt.title("Male vs Female Employees")
plt.show()
```

Experiment 1.5: Draw a scatter plot showing the relationship between experience and salary.

```python
# Scatter plot between experience and salary
plt.figure(figsize=(8,5))

plt.scatter(df["Experience_Years"],
            df["Salary"])

plt.title("Experience vs Salary")
plt.xlabel("Experience (Years)")
plt.ylabel("Salary")
plt.show()
```

Experiment 1.6: Display the top 10 employees with the highest salary.

```python
# Sort data by salary in descending order
top20 = df.sort_values(by="Salary", ascending=False).head(20)

print("Top 20 Highest Paid Employees")
print(top20)
```

Experiment 1.7: Find the highest salary in every department.

```python
# Maximum salary department-wise
highest_salary = df.groupby("Department")["Salary"].max()

print("Highest Salary in Each Department")
print(highest_salary)
```

Experiment 1.8: Display employees whose salary is greater than the overall average salary.

```python
# Calculate overall average salary
average_salary = df["Salary"].mean()

print("Overall Average Salary:", average_salary)

# Employees earning above average salary
above_average = df[df["Salary"] > average_salary]

print(above_average)
```

Experiment 1.9: Calculate the average years of experience in each department.

```python
# Average experience department-wise
avg_experience = df.groupby("Department")["Experience_Years"].mean()

print("Average Experience in Each Department")
print(avg_experience)
```

Experiment 1.10: Display the distribution of employee ages using a histogram.

```python
# Histogram of employee ages
plt.figure(figsize=(8,5))

plt.hist(df["Age"],
         bins=10,
         color="purple",
         edgecolor="black")

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of Employees")
plt.show()
```
