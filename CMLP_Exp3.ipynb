Experiment 3.1: Stemming using Porter Stemmer

```python
from nltk.stem import PorterStemmer

# Create a Porter Stemmer object
stemmer = PorterStemmer()
```

```python
# Words given in the experiment
words = [
    "playing",
    "played",
    "plays",
    "studies",
    "studying",
    "connected",
    "connection",
    "computers"
]

# Apply stemming to each word
for word in words:
    stem = stemmer.stem(word)
    print(f"Original: {word}  ->  Stem: {stem}")
```

Experiment 3.2: Lemmatization

```python
from nltk.stem import WordNetLemmatizer

# Create a WordNet Lemmatizer object
lemmatizer = WordNetLemmatizer()
```

```python
# Words given in the experiment
words = [
    "cats",
    "dogs",
    "running",
    "runs",
    "ran",
    "studies",
    "studying",
    "better",
    "children",
    "mice",
    "went",
    "ate",
    "leaves",
    "caring"
]

# Apply lemmatization to each word
for word in words:
    lemma = lemmatizer.lemmatize(word)
    print(f"Original: {word}  ->  Lemma: {lemma}")
```

```python
# Lemmatize words by specifying their grammatical category
words_with_pos = [
    ("cats", "n"),
    ("dogs", "n"),
    ("running", "v"),
    ("runs", "v"),
    ("ran", "v"),
    ("studies", "v"),
    ("studying", "v"),
    ("better", "a"),
    ("children", "n"),
    ("mice", "n"),
    ("went", "v"),
    ("ate", "v"),
    ("leaves", "n"),
    ("caring", "v")
]

# Display original word and its lemma
for word, pos in words_with_pos:
    lemma = lemmatizer.lemmatize(word, pos=pos)
    print(f"Original: {word}  ->  Lemma: {lemma}")
```

```python
import re

# Input text given in the experiment
text = """
Welcome to the Natural Language Processing workshop! For more information,
visit https://www.nlpworkshop.com or www.python.org. You can contact the
coordinator at nlpworkshop@gmail.com or support@python.org. For registration,
call +91-9876543210 or 9123456789. Follow us on social media @NLPWorkshop
and @PythonLearner. Share your experience using #NLP, #Python, and
#MachineLearning. You can also visit https://github.com/NLPWorkshop
for the latest updates.
"""
```

```python
# Regular expression pattern for email addresses
email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'

emails = re.findall(email_pattern, text)

print("Email Addresses:")
for email in emails:
    print(email)
```

```python
# Regular expression pattern for URLs
url_pattern = r'https?://[^\s]+|www\.[^\s]+'

urls = re.findall(url_pattern, text)

print("URLs:")
for url in urls:
    print(url)
```

```python
# Regular expression pattern for Indian mobile numbers
mobile_pattern = r'(?:\+91[-\s]?)?[6-9]\d{9}'

mobile_numbers = re.findall(mobile_pattern, text)

print("Mobile Numbers:")
for number in mobile_numbers:
    print(number)
```

```python
# Regular expression pattern for hashtags
hashtag_pattern = r'#\w+'

hashtags = re.findall(hashtag_pattern, text)

print("Hashtags:")
for hashtag in hashtags:
    print(hashtag)
```

```python
# Regular expression pattern for social media mentions
mention_pattern = r'@\w+'

mentions = re.findall(mention_pattern, text)

print("Mentions:")
for mention in mentions:
    print(mention)
```

```python
# Display all extracted information together

print("===== EMAIL ADDRESSES =====")
print(emails)

print("\n===== URLS =====")
print(urls)

print("\n===== MOBILE NUMBERS =====")
print(mobile_numbers)

print("\n===== HASHTAGS =====")
print(hashtags)

print("\n===== MENTIONS =====")
print(mentions)
```
